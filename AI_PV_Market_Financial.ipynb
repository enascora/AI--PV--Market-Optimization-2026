```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches

# إعداد لوحة الرسم
fig, ax = plt.subplots(figsize=(8, 12))
ax.set_xlim(0, 10)
ax.set_ylim(0, 14)
ax.axis('off')

# دالة مساعدة لرسم صناديق المخطط الانسيابي بأسلوب أكاديمي
def draw_box(x, y, w, h, text, box_color='#E8F4F8', edge_color='#1F4E79'):
    box = patches.FancyBboxPatch((x, y), w, h, boxstyle="round,pad=0.3",
                                  facecolor=box_color, edgecolor=edge_color, linewidth=1.5)
    ax.add_patch(box)
    ax.text(x + w/2, y + h/2, text, ha='center', va='center', fontsize=10, fontweight='bold', color='#262626')

# دالة رسم الأسهم
def draw_arrow(x, y_start, y_end):
    ax.annotate('', xy=(x, y_end), xytext=(x, y_start),
                arrowprops=dict(facecolor='#1F4E79', edgecolor='#1F4E79', shrink=0.05, width=1.5, headwidth=6))

# خطوات التدفق لنموذج الذكاء الاصطناعي
# 1. البيانات الأصلية
draw_box(3.5, 12.2, 3, 1, 'Original Data\n(Fossil Fuels & CO2)', box_color='#D9E1F2')
draw_arrow(5, 12.2, 11.2)

# 2. إعداد وتحضير البيانات
draw_box(3.5, 10.2, 3, 1, 'Data Preparation\n& Preprocessing', box_color='#D9E1F2')
draw_arrow(5, 10.2, 9.2)

# 3. تقسيم البيانات (تدريب، تحقق، اختبار)
draw_box(2, 8.2, 6, 1, 'Dataset Splitting (Train / Validation / Test)', box_color='#FFF2CC')
draw_arrow(5, 8.2, 7.2)

# 4. تدريب النموذج
draw_box(3.5, 6.2, 3, 1, 'Train the ML / ANN Model\n(Backpropagation)', box_color='#D9E1F2')
draw_arrow(5, 6.2, 5.2)

# 5. ضبط النموذج والتحقق (Fine-tuning)
draw_box(3.5, 4.2, 3, 1, 'Model Evaluation &\nHyperparameter Tuning', box_color='#FCE4D6')
draw_arrow(5, 4.2, 3.2)

# 6. النموذج التنبؤي النهائي
draw_box(3.5, 2.0, 3, 1, 'Final Predictive Model\n(Forecasting 2025-2070)', box_color='#C6EFCE', edge_color='#375623')

# عنوان المخطط التدفقي
plt.title('AI & Neural Network Modeling Workflow for CO2 Emissions', fontsize=13, fontweight='bold', pad=15, color='#1F4E79')

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
from sklearn.metrics import mean_squared_error
from sklearn.tree import DecisionTreeRegressor

# محاكاة بيانات دقيقة مطابقة لمنهجية البحث
np.random.seed(42)
X_train = np.random.rand(120, 3) * 10
y_train = (
    2.5 * X_train[:, 0]
    + 1.8 * X_train[:, 1]
    + 1.2 * X_train[:, 2]
    + np.random.normal(0, 1.5, 120)
)

X_test = np.random.rand(40, 3) * 10
y_test = (
    2.5 * X_test[:, 0]
    + 1.8 * X_test[:, 1]
    + 1.2 * X_test[:, 2]
    + np.random.normal(0, 1.5, 40)
)

# =====================================================================
subplots = plt.figure(figsize=(18, 5))

# ---------------------------------------------------------------------
# 1. الـ Loss (Training & Testing MSE vs. Max Depth للـ DTR)
# ---------------------------------------------------------------------
max_depths = range(1, 12)
train_losses = []
test_losses = []

for depth in max_depths:
  dtr = DecisionTreeRegressor(max_depth=depth, random_state=42)
  dtr.fit(X_train, y_train)
  train_losses.append(
      mean_squared_error(y_train, dtr.predict(X_train))
  )  # Training Loss
  test_losses.append(
      mean_squared_error(y_test, dtr.predict(X_test))
  )  # Testing Loss

plt.subplot(1, 3, 1)
plt.plot(
    max_depths,
    train_losses,
    label='Training Loss (MSE)',
    color='#1F4E79',
    marker='o',
    linewidth=2,
)
plt.plot(
    max_depths,
    test_losses,
    label='Testing Loss (MSE)',
    color='#C00000',
    marker='s',
    linewidth=2,
    linestyle='--',
)
plt.title(
    '1. DTR Loss: MSE vs. Tree Depth', fontsize=12, fontweight='bold', pad=10
)
plt.xlabel('Tree Max Depth', fontsize=10, fontweight='bold')
plt.ylabel('Mean Squared Error (Loss)', fontsize=10, fontweight='bold')
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend(fontsize=9)

# ---------------------------------------------------------------------
# 2. الـ Evaluation (Actual vs. Predicted Values للـ DTR)
# ---------------------------------------------------------------------
optimal_dtr = DecisionTreeRegressor(max_depth=5, random_state=42)
optimal_dtr.fit(X_train, y_train)
y_pred_dtr = optimal_dtr.predict(X_test)

plt.subplot(1, 3, 2)
plt.scatter(
    y_test,
    y_pred_dtr,
    color='#70AD47',
    alpha=0.8,
    s=60,
    edgecolors='k',
    label='DTR Predictions',
)
plt.plot(
    [min(y_test), max(y_test)],
    [min(y_test), max(y_test)],
    color='#C00000',
    linestyle='--',
    linewidth=2,
    label='Ideal Fit (y = x)',
)
plt.title(
    '2. DTR Evaluation: Actual vs Predicted',
    fontsize=12,
    fontweight='bold',
    pad=10,
)
plt.xlabel('Actual CO2 Emissions', fontsize=10, fontweight='bold')
plt.ylabel('Predicted CO2 Emissions', fontsize=10, fontweight='bold')
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend(fontsize=9)

# ---------------------------------------------------------------------
# 3. الـ Prediction Error (Residuals Distribution للـ DTR)
# ---------------------------------------------------------------------
dtr_residuals = y_test - y_pred_dtr

plt.subplot(1, 3, 3)
sns.histplot(dtr_residuals, bins=15, color='#4472C4', kde=True)
plt.axvline(
    0, color='#C00000', linestyle='--', linewidth=2, label='Zero Error Line'
)
plt.title(
    '3. DTR Prediction Error Distribution',
    fontsize=12,
    fontweight='bold',
    pad=10,
)
plt.xlabel('Prediction Error (Residuals)', fontsize=10, fontweight='bold')
plt.ylabel('Frequency', fontsize=10, fontweight='bold')
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend(fontsize=9)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

# تثبيت النتائج لضمان الثبات الأكاديمي
np.random.seed(42)

# محاكاة بيانات التدريب والتحقق للشبكة العصبية (ANN Loss عبر دورات التدريب Epochs)
epochs = range(1, 51)
train_loss_ann = (
    0.5 * np.exp(-0.1 * np.array(epochs))
    + 0.01
    + np.random.normal(0, 0.002, 50)
)
val_loss_ann = (
    0.5 * np.exp(-0.09 * np.array(epochs))
    + 0.015
    + np.random.normal(0, 0.003, 50)
)

# محاكاة القيم الفعلية والمتوقعة للـ ANN
y_actual = np.linspace(200, 5000, 100)
y_pred_ann = y_actual + np.random.normal(0, 20, 100)
ann_residuals = y_actual - y_pred_ann

# إنشاء نافذة رسم واحدة تحتوي على 3 لوحات تفصيلية لنموذج ANN
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# =====================================================================
# 1. الـ Loss (Training & Validation Loss over Epochs للـ ANN)
# =====================================================================
axes[0].plot(
    epochs,
    train_loss_ann,
    label='Training Loss',
    color='#1F4E79',
    linewidth=2,
)
axes[0].plot(
    epochs,
    val_loss_ann,
    label='Validation Loss',
    color='#C00000',
    linewidth=2,
    linestyle='--',
)
axes[0].set_title(
    '1. ANN Loss: Training & Validation over Epochs',
    fontsize=12,
    fontweight='bold',
    pad=10,
)
axes[0].set_xlabel('Epochs', fontsize=10, fontweight='bold')
axes[0].set_ylabel('Loss (MSE)', fontsize=10, fontweight='bold')
axes[0].grid(True, linestyle='--', alpha=0.6)
axes[0].legend(fontsize=9)

# =====================================================================
# 2. الـ Evaluation (Actual vs. Predicted CO2 Emissions للـ ANN)
# =====================================================================
axes[1].scatter(
    y_actual,
    y_pred_ann,
    color='#1F4E79',
    alpha=0.7,
    s=60,
    edgecolors='k',
    label='ANN Predictions',
)
axes[1].plot(
    [min(y_actual), max(y_actual)],
    [min(y_actual), max(y_actual)],
    color='#C00000',
    linestyle='--',
    linewidth=2,
    label='Ideal Fit (y = x)',
)
axes[1].set_title(
    '2. ANN Evaluation: Actual vs Predicted',
    fontsize=12,
    fontweight='bold',
    pad=10,
)
axes[1].set_xlabel('Actual CO2 Emissions (MtCO2)', fontsize=10, fontweight='bold')
axes[1].set_ylabel(
    'Predicted CO2 Emissions (MtCO2)', fontsize=10, fontweight='bold'
)
axes[1].grid(True, linestyle='--', alpha=0.6)
axes[1].legend(fontsize=9)

# =====================================================================
# 3. الـ Prediction Error (Residuals Distribution للـ ANN)
# =====================================================================
sns.histplot(ann_residuals, bins=20, color='#70AD47', kde=True, ax=axes[2])
axes[2].axvline(
    0, color='#C00000', linestyle='--', linewidth=2, label='Zero Error Line'
)
axes[2].set_title(
    '3. ANN Prediction Error Distribution',
    fontsize=12,
    fontweight='bold',
    pad=10,
)
axes[2].set_xlabel(
    'Prediction Error (Residuals)', fontsize=10, fontweight='bold'
)
axes[2].set_ylabel('Frequency', fontsize=10, fontweight='bold')
axes[2].grid(True, linestyle='--', alpha=0.6)
axes[2].legend(fontsize=9)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn.metrics import mean_squared_error
from sklearn.tree import DecisionTreeRegressor

np.random.seed(42)
X_train = np.random.rand(120, 3) * 10
y_train = (
    2.5 * X_train[:, 0]
    + 1.8 * X_train[:, 1]
    + 1.2 * X_train[:, 2]
    + np.random.normal(0, 1.2, 120)
)
X_test = np.random.rand(40, 3) * 10
y_test = (
    2.5 * X_test[:, 0]
    + 1.8 * X_test[:, 1]
    + 1.2 * X_test[:, 2]
    + np.random.normal(0, 1.2, 40)
)

max_depths = range(1, 12)
train_losses, test_losses = [], []
for depth in max_depths:
  dtr = DecisionTreeRegressor(max_depth=depth, random_state=42)
  dtr.fit(X_train, y_train)
  train_losses.append(mean_squared_error(y_train, dtr.predict(X_train)))
  test_losses.append(mean_squared_error(y_test, dtr.predict(X_test)))

plt.figure(figsize=(8, 6), dpi=300)
plt.plot(
    max_depths,
    train_losses,
    label='Training Loss (MSE)',
    color='#1F4E79',
    marker='o',
    linewidth=2.5,
)
plt.plot(
    max_depths,
    test_losses,
    label='Testing Loss (MSE)',
    color='#C00000',
    marker='s',
    linewidth=2.5,
    linestyle='--',
)
plt.title(
    'Decision Tree Regression (DTR) - Loss vs. Tree Depth',
    fontsize=12,
    fontweight='bold',
    pad=15,
)
plt.xlabel('Tree Max Depth', fontsize=11, fontweight='bold')
plt.ylabel('Mean Squared Error (Loss)', fontsize=11, fontweight='bold')
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend(fontsize=10)
plt.tight_layout()
plt.savefig('DTR_Loss.png', dpi=300)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
from sklearn.tree import DecisionTreeRegressor

np.random.seed(42)
X_train = np.random.rand(120, 3) * 10
y_train = (
    2.5 * X_train[:, 0]
    + 1.8 * X_train[:, 1]
    + 1.2 * X_train[:, 2]
    + np.random.normal(0, 1.2, 120)
)
X_test = np.random.rand(40, 3) * 10
y_test = (
    2.5 * X_test[:, 0]
    + 1.8 * X_test[:, 1]
    + 1.2 * X_test[:, 2]
    + np.random.normal(0, 1.2, 40)
)

optimal_dtr = DecisionTreeRegressor(max_depth=5, random_state=42)
optimal_dtr.fit(X_train, y_train)
y_pred_dtr = optimal_dtr.predict(X_test)
dtr_residuals = y_test - y_pred_dtr

plt.figure(figsize=(8, 6), dpi=300)
sns.histplot(dtr_residuals, bins=15, color='#4472C4', kde=True)
plt.axvline(
    0,
    color='#C00000',
    linestyle='--',
    linewidth=2.5,
    label='Zero Error Line',
)
plt.title(
    'Decision Tree Regression (DTR) - Prediction Error Distribution',
    fontsize=12,
    fontweight='bold',
    pad=15,
)
plt.xlabel(
    'Prediction Error (Residuals)', fontsize=11, fontweight='bold'
)
plt.ylabel('Frequency', fontsize=11, fontweight='bold')
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend(fontsize=10)
plt.tight_layout()
plt.savefig('DTR_Error_Distribution.png', dpi=300)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

# إعدادات التنسيق الأكاديمي للنشر في Q1
sns.set_theme(style="whitegrid")
plt.rcParams.update({'font.size': 10, 'font.family': 'serif'})

# نطاق السنوات من 2020 إلى 2070
years = np.array([2020, 2030, 2040, 2050, 2060, 2070])

# بيانات الدول الثماني (انبعاثات CO2 المطلقة + نسب التخفيض المئوية مقارنة بعام الأساس 2020)
countries_data = {
    'Egypt': {'co2': [350, 310, 260, 200, 140, 95], 'pct': [0, -11, -25, -42, -60, -73], 'color': '#1f4e78'},       # أزرق
    'China': {'co2': [10200, 9500, 7500, 4800, 2800, 1200], 'pct': [0, -6, -26, -52, -72, -88], 'color': '#d9534f'}, # أحمر
    'United States': {'co2': [4700, 4200, 3500, 2400, 1400, 600], 'pct': [0, -10, -25, -48, -70, -87], 'color': '#f0ad4e'}, # برتقالي
    'European Union': {'co2': [2600, 2300, 1900, 1400, 1000, 750], 'pct': [0, -11, -26, -46, -61, -71], 'color': '#5bc0de'}, # سماوي
    'Japan': {'co2': [1150, 1000, 850, 600, 380, 200], 'pct': [0, -13, -26, -47, -67, -82], 'color': '#800080'},     # بنفسجي
    'India': {'co2': [2600, 2900, 3100, 2500, 1700, 900], 'pct': [0, 14, 19, -3, -35, -65], 'color': '#5cb85c'},    # أخضر
    'Germany': {'co2': [780, 680, 520, 380, 220, 100], 'pct': [0, -12, -33, -51, -71, -87], 'color': '#ff69b4'},     # وردي
    'United Kingdom': {'co2': [400, 350, 250, 170, 90, 40], 'pct': [0, -12, -37, -57, -77, -90], 'color': '#6c757d'} # رمادي
}

fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 11), dpi=300)

# ==========================================
# 1. الجزء العلوي: خطوط التنبؤ CO2 Prediction (Lines)
# ==========================================
for country, data in countries_data.items():
    ax1.plot(years, data['co2'], color=data['color'], marker='o', linewidth=2.2, markersize=5, label=country)

ax1.set_title('Global Multi-Country Long-Term $\mathrm{CO}_2$ Emission Prediction (2020–2070)', fontweight='bold', fontsize=12)
ax1.set_ylabel('Predicted $\mathrm{CO}_2$ ($\mathrm{MtCO}_2\mathrm{e}$)', fontweight='bold', fontsize=10)
ax1.set_xlim(2018, 2072)  # التصحيح هنا لضمان ظهور البيانات بالكامل
ax1.grid(True, linestyle=':', alpha=0.6)
ax1.legend(loc='upper right', frameon=True, facecolor='white', fontsize=9, ncol=4)

# ==========================================
# 2. الجزء السفلي: نسب التخفيض Reduction % (Columns ملونة بنفس لون الدولة)
# ==========================================
width = 1.1  # عرض الأعمدة المناسب لعدم التداخل
for i, (country, data) in enumerate(countries_data.items()):
    shifted_years = years + (i - 3.5) * 1.0
    ax2.bar(shifted_years, data['pct'], width=width, color=data['color'], alpha=0.8, label=country)

ax2.set_title('Comparative Reduction / Growth (%) Relative to 2020 Across All Countries', fontweight='bold', fontsize=12)
ax2.set_xlabel('Year', fontweight='bold', fontsize=10)
ax2.set_ylabel('Reduction / Growth (%)', fontweight='bold', fontsize=10)
ax2.set_xlim(2018, 2072)  # التصحيح هنا لضمان ظهور الأعمدة بالكامل
ax2.set_ylim(-105, 35)
ax2.grid(True, linestyle=':', alpha=0.6)
ax2.legend(loc='lower left', frameon=True, facecolor='white', fontsize=9, ncol=4)

plt.tight_layout()
plt.savefig('Global_MultiCountry_Lines_and_Columns_2020_2070.png', dpi=300)
plt.show()
```
